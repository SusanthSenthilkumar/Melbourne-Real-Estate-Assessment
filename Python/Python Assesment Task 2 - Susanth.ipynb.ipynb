```python
df['price_per_room'] = df['price'] / df['rooms']
```

```python
df['price_per_land'] = df['price'] / df['landsize'].replace(0, np.nan)
```

```python
df['property_age'] = 2017 - df['yearBuilt']
```

```python
df['yearBuilt'] = df['yearBuilt'].fillna(df['yearBuilt'].median())
df['property_age'] = 2017 - df['yearBuilt']
```

New features were created to enhance property-level analysis:

• Price per Room: Calculated as price / rooms to evaluate cost efficiency of each property
• Price per Land Square Meter: Calculated as price / landsize to understand land value contribution
