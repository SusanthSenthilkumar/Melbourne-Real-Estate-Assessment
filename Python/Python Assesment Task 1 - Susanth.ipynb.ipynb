```python
import pandas as pd
import numpy as np

df = pd.read_csv("C:/Users/susan/Downloads/assessment_melb vserve.xlsx.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df['buildingArea'] = df['buildingArea'].fillna(df['buildingArea'].median())
df['yearBuilt'] = df['yearBuilt'].fillna(df['yearBuilt'].median())
```

```python
Q1 = df['price'].quantile(0.25)
Q3 = df['price'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
```

```python
df = df[(df['price'] >= lower) & (df['price'] <= upper)]
```

```python
df.shape
```

Median was chosen over mean as it is more robust to extreme values commonly present in real estate data.
Outliers were removed using the IQR method to ensure extreme property prices do not skew analysis.
