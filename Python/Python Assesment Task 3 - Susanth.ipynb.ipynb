```python
import pandas as pd
import numpy as np

df = pd.read_csv("C:/Users/susan/Downloads/assessment_melb vserve.xlsx.csv")
```

```python
df.head()
```

```python
df['buildingArea'] = df['buildingArea'].fillna(df['buildingArea'].median())
```

```python
suburb_df = df.groupby('suburb').agg({
    'price': 'mean',
    'buildingArea': 'mean'
}).reset_index()
```

```python
suburb_df['price_building_ratio'] = suburb_df['price'] / suburb_df['buildingArea']
```

```python
suburb_df = suburb_df.sort_values(by='price_building_ratio', ascending=False)
```

Task 3: Suburb-Level Feature Engineering

The ratio of average price to average building area was calculated for each suburb.

Steps:
• Grouped data by suburb
• Calculated average price and average building area
• Computed ratio = avg(price) / avg(buildingarea)

Missing values in buildingarea were handled using median to ensure accurate aggregation.

This metric helps compare property value relative to built-up space across suburbs.
